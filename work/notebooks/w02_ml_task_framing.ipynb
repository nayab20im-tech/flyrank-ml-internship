## 1. My lane as an ML task (type)

**Unsupervised Clustering**

**Why:** Rather than predicting a simple binary label, my freestyle goal is to discover natural "performance archetypes" across the site based on multidimensional behavior (impressions, positions, CTR, age, and word count). This allows us to assign tailored SEO strategies to each group instead of treating every page the same.

## 2. Target or proxy

**Target/Proxy:** Because clustering is unsupervised, there is no single target column. Instead, the assigned **Cluster ID** serves as a proxy for the page's current performance state (e.g., Cluster 0 = High Impression / Low CTR, Cluster 1 = Stale Content Decay, Cluster 2 = High Performing Core).

## 3. Success metric

**Success Metric:**

  **Statistical:** Silhouette Score (measuring how distinct and well-separated the clusters are).

  **Business Actionability:** Whether each cluster maps cleanly to a distinct, actionable SEO play (e.g., "title tag rewrite" vs. "full content rewrite").

## 5. Why ML beats a fixed rule here

Human rules use rigid 1D or 2D thresholds (like CTR < 1%). They struggle to draw boundaries across 5+ interacting dimensions simultaneously. Clustering algorithms automatically discover non-linear patterns in high-dimensional space that an if-else statement would miss.
